## Task 1: Create a List of Song IDs

```python
playlist_ids = [101, 102, 103, 104, 105]

print("Initial Playlist:", playlist_ids)
```

## Task 2: Add IDs Using append() and extend()

```python

playlist_ids.append(106)


playlist_ids.extend([107])

print("Updated Playlist:", playlist_ids)
```

## Task 3: Remove the Last Song Using pop()

```python

removed_song = playlist_ids.pop()

print("Removed Song ID:", removed_song)
print("Remaining Playlist:", playlist_ids)
```

## Task 5: Choosing Between List and Tuple

```python

zomato_recent_orders = ["Paneer Tikka", "Cold Coffee", "Veg Biryani"]


ipl_teams = ("CSK", "MI", "RCB", "KKR", "GT")

print("Zomato orders should use:", type(zomato_recent_orders).__name__, "(mutable, frequently updated)")
print("IPL teams should use:", type(ipl_teams).__name__, "(immutable, constant collection)")
```

## Task 1: Lowercase & Replace Dash

```python
text = 'Flipkart-Sale2024'
result = text.lower().replace('-', ' ')

print(result)
```

## Task 2: Clean Product Name

```python
raw_name = '  OnePlus Nord-CE 3  '
cleaned_name = raw_name.strip().upper().replace('-', ':')

print(cleaned_name)
```

## Task 3: Split Product Code Function

```python
def split_product_code(product_code):
    return product_code.split('-')


code = 'ZOMATO-FOOD-2024'
parts = split_product_code(code)

print(parts)
```

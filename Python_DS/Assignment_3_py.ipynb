## Task 2: Cart Total Function (String to Float)

```python
def total_cart_amount(prices):
    total = 0.0
    for price in prices:
        total += float(price)
    return total

# sample flipkart cart
cart_items = ['199.99', '49', '350.75']
bill = total_cart_amount(cart_items)

print("Total Cart Amount:", bill)
```

## Task 4: String to Boolean Conversion

```python
is_premium = 'True'

# converting by checking equality
actual_bool = is_premium.lower() == 'true'

print("Value:", actual_bool)
print("Type:", type(actual_bool))
```
